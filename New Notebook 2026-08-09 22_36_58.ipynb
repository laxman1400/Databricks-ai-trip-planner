```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()

results = w.vector_search_indexes.query_index(
    index_name="workspace.trip_planner_latest.attraction_search",
    query_text="scenic outdoor places good for photography and an easy walk",
    columns=[
        "document_id",
        "destination",
        "title",
        "content",
        "source_url"
    ],
    num_results=5,
    query_type="HYBRID"
)

print(results)
```
